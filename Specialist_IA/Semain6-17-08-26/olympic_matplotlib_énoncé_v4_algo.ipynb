```python
# This block imports the libraries.
# Seaborn needs a pandas DataFrame, so we convert your dictionary into a DataFrame.

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame(data)


# This block sets the Seaborn theme and removes rows where Height or Sex is missing.

sns.set_theme(style="whitegrid")

df_height = df.dropna(subset=["Height", "Sex"])


# This block creates height groups of 4 cm.
# The same bins are used for both men and women.

height_min = int(df_height["Height"].min())
height_max = int(df_height["Height"].max())

bins = range(height_min, height_max + 5, 4)


# This block creates the Seaborn histogram.
# hue="Sex" separates men and women automatically.
# palette gives the requested colors.

plt.figure(figsize=(12, 6))

sns.histplot(
    data=df_height,
    x="Height",
    hue="Sex",
    bins=bins,
    multiple="layer",
    palette={"M": "steelblue", "F": "pink"},
    alpha=0.6,
    edgecolor="black"
)

plt.xlabel("Height")
plt.ylabel("Nombre d'athlètes")
plt.title("Distribution des tailles par sexe et par tranche de 4 cm")

plt.show()

# This block creates two dictionaries:
# one for men's unique sports per year,
# one for women's unique sports per year.

sports_par_annee_hommes = {}
sports_par_annee_femmes = {}

for year, sport, sex in zip(data["Year"], data["Sport"], data["Sex"]):
    if sex == "M":
        if year not in sports_par_annee_hommes:
            sports_par_annee_hommes[year] = set()
        sports_par_annee_hommes[year].add(sport)

    elif sex == "F":
        if year not in sports_par_annee_femmes:
            sports_par_annee_femmes[year] = set()
        sports_par_annee_femmes[year].add(sport)

        # This block counts the number of unique sports per year for men and women.

nombre_sports_hommes = {
    year: len(sports)
    for year, sports in sports_par_annee_hommes.items()
}

nombre_sports_femmes = {
    year: len(sports)
    for year, sports in sports_par_annee_femmes.items()
}


# This block creates a clean DataFrame for Seaborn.
# Seaborn works better with long-format data:
# Year | Genre | Nombre_sports

years = sorted(set(nombre_sports_hommes.keys()) | set(nombre_sports_femmes.keys()))

rows = []

for year in years:
    rows.append({
        "Year": year,
        "Genre": "Hommes",
        "Nombre_sports": nombre_sports_hommes.get(year, 0)
    })

    rows.append({
        "Year": year,
        "Genre": "Femmes",
        "Nombre_sports": nombre_sports_femmes.get(year, 0)
    })

df_sports = pd.DataFrame(rows)

# This block creates the Seaborn line plot.
# marker="o" adds points.
# linestyle="--" makes the line dashed.
# palette applies cyan and magenta.
# alpha=0.5 makes the lines semi-transparent.

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=df_sports,
    x="Year",
    y="Nombre_sports",
    hue="Genre",
    marker="o",
    linestyle="--",
    palette={
        "Hommes": "magenta",
        "Femmes": "cyan"
    },
    alpha=0.5
)

plt.xlabel("Année")
plt.ylabel("Nombre de sports uniques")
plt.title("Nombre de sports uniques par année et par genre")

plt.legend(loc="lower left")

plt.show()
```

5. __Faire un bar-plot qui montre:__<br>
    - le top 10 des noms 
    - qui ont participé au plus de sports différents, 
    - trié par: nombre de sports différents, nombre de médailles
    - changer l'orientation des labels sur l'axe-x ('xticks')
    - changer le nom de l'axe-x: 'top 10' ('xlabel')
<br>

```python
# This block builds one dictionary per athlete.
# Key = athlete ID
# Value = name, set of unique sports, and medal counter.

athletes = {}

for athlete_id, name, sport, medal in zip(data["ID"], data["Name"], data["Sport"], data["Medal"]):
    if athlete_id not in athletes:
        athletes[athlete_id] = {
            "name": name,
            "sports": set(),
            "medals": 0
        }

    # Add the sport to a set, so duplicates are automatically ignored.
    athletes[athlete_id]["sports"].add(sport)

    # Count the medal only if the Medal column is not empty / NaN.
    if pd.notna(medal):
        athletes[athlete_id]["medals"] += 1

# This block converts the athlete dictionary into a list of result rows.
# Each row contains the name, the number of different sports, and the number of medals.

resultats = []

for athlete_id, info in athletes.items():
    resultats.append({
        "ID": athlete_id,
        "Name": info["name"],
        "Nombre_sports": len(info["sports"]),
        "Nombre_medailles": info["medals"]
    })

# This block sorts the athletes.
# First criterion: number of different sports.
# Second criterion: number of medals.
# reverse=True puts the biggest values first.

resultats_tries = sorted(
    resultats,
    key=lambda athlete: (athlete["Nombre_sports"], athlete["Nombre_medailles"]),
    reverse=True
)


# This block keeps only the top 10 athletes.

top_10 = resultats_tries[:10]

# This block prepares the values for the bar plot.
# x = athlete names
# y = number of different sports

noms = [athlete["Name"] for athlete in top_10]
nombre_sports = [athlete["Nombre_sports"] for athlete in top_10]
nombre_medailles = [athlete["Nombre_medailles"] for athlete in top_10]

# This block prints the top 10 with sports and medals.
# It helps verify that the sorting is correct.

for athlete in top_10:
    print(
        athlete["Name"],
        "- Sports différents:",
        athlete["Nombre_sports"],
        "- Médailles:",
        athlete["Nombre_medailles"]
    )
# This block creates the bar plot.
# The height of each bar represents the number of different sports.

plt.bar(noms, nombre_sports)

plt.xlabel("top 10")
plt.ylabel("Nombre de sports différents")
plt.title("Top 10 des athlètes ayant participé au plus de sports différents")

plt.xticks(rotation=45, ha="right")

plt.show()
```

```python
# This full block creates a pie plot for Carl Schuhmann.
# It shows the number of medals per sport and highlights the sport with the most medals.

import pandas as pd
import matplotlib.pyplot as plt

nom_recherche = "Carl Schuhmann"

ids_trouves = set()

for athlete_id, name in zip(data["ID"], data["Name"]):
    if nom_recherche.lower() in name.lower():
        ids_trouves.add(athlete_id)
        print(athlete_id, "-", name)

athlete_id_choisi = list(ids_trouves)[0]

medailles_par_sport = {}

for athlete_id, sport, medal in zip(data["ID"], data["Sport"], data["Medal"]):
    if athlete_id == athlete_id_choisi and pd.notna(medal):
        if sport not in medailles_par_sport:
            medailles_par_sport[sport] = 0

        medailles_par_sport[sport] += 1

sports = list(medailles_par_sport.keys())
nombre_medailles = list(medailles_par_sport.values())

max_medailles = max(nombre_medailles)

explode = [
    0.1 if nb == max_medailles else 0
    for nb in nombre_medailles
]

plt.pie(
    nombre_medailles,
    labels=sports,
    autopct="%1.1f%%",
    explode=explode,
    startangle=90
)

plt.title("Médailles de Carl Schuhmann par sport")

plt.show()
```
