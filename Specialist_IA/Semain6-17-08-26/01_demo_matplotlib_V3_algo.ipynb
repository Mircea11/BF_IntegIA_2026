## data

```python
iris = sns.load_dataset("iris") ## pd.read_csv("iris.csv")
iris.sample(8)

iris = iris.to_dict(orient="list")
iris.keys()
```

```python
poly = {}
poly["x"] = [i/4 for i in range(40)]
poly["y1"] = [(x/5 - 0.5) for x in poly["x"]]#poly["x"]/5 - 0.5
poly["y2"] = [y**2 for y in poly["y1"]] #poly["y1"]**2
display(poly)
```

```python
GB1 = {col:[] for col in iris.keys()}
sub_data = {spe:{col:[] for col in iris.keys()} for spe in set(iris["species"])}
for spe in set(iris["species"]):
    for i,s in enumerate(iris["species"]):
        if s == spe:
            for col in iris.keys():
                
                sub_data[spe][col].append(iris[col][i])
    
    for col in ["petal_length","petal_width","sepal_length","sepal_width"]:
        content_list = sub_data[spe][col]
        GB1[col].append(sum(content_list)/len(content_list))
    GB1["species"].append(spe)
```

## histogram
- histogramme
- nécessite:
    - x en numérique
    - (bins) : compartiments/découpes des valeurs de 'x' en catégories contigu
- compte le nombre d'occurrences/lignes pour lequel les valeurs de 'x' se retrouve dans ces compartiments


```python
plt.hist(x=[1.1,2.1,2.2,3.1,3.2,3.3,4.1,4.2,4.3,4.4,4.5]
         ,bins=[-.5,.5,1.5,2.5,3.5,4.5,5.5,6.5] #range(0,7,1) ## de 0 (inclu) à 1 exclu, 1 (inclu) à 2 (exclu), ...
        )
plt.show()
```

```python
bins = [(i/2+0.25) for i in range(16)]

plt.hist(data = iris 
         , x='petal_length'
         , bins=bins#np.linspace(0,10,26)+0.25
    )
plt.show()
```

## pie
- graphique en secteurs, camembert
- simmilaire au 'barplot', mais vraiment utile que pour afficher des proportions
- nécessite:
    - x : ce qui indique la taille des secteur (proportionnel)
    - (label): nom des secteurs
    - (autopct): format des pourcentage à utiliser
        - exemple:
            - "%" + "1.2f" + "%%" = afficher 2 chiffes après la virgule
            - "%" + "1.0f" + "%%" = afficher 0 chiffes après la virgule

```python
plt.pie(data=GB1, x='petal_length', labels='species', autopct='%1.2f%%')
plt.show()
```

```python
plt.pie(data=GB1, x='petal_length', labels='species', autopct='%1.0f%%'
        ,explode=(0.2,0.1,0) ## mise en évidence de la premiere entrée= decrochage du secteur
       ) 
plt.show()
```

# Superposing
- Superposition des couches de graphiques

```python

x = poly["x"]
y1 = poly["y1"]
y2 = poly["y2"]

plt.plot(x, y1, label='y1') ## label: nommer la couche
plt.plot(x, y2, label='y2') ## ajouter une couche par dessus la couche précédente

plt.legend() ## legende : afficher le nom de la couche (par défaut= dans le meilleur emplacement possible du graphique)
plt.show()
```

```python

x = poly["x"]
y1 = poly["y1"]
y2 = poly["y2"]

plt.plot(x, y1, label='y1')
plt.plot(x, y2, label='y2') 

plt.legend(loc = "lower right" ## imposer l'emplacement de la légende
           , ncol=2   ## imposer 2 colonnes
           )
plt.show()
```

```python
###

setosa = sub_data["setosa"]
versicolor = sub_data["versicolor"]
virginica = sub_data["virginica"]


bins = [i/2+0.25 for i in range(0,17)] ##np.linspace(0,8,17)

plt.hist(data=sub_data["setosa"],x="petal_length"
         ,bins=bins
         ,alpha=0.5 ## ajouter de la transparence pour voir la supperposition
         ,label="setosa"
        ) 
plt.hist(data=sub_data["versicolor"],x="petal_length",bins=bins,alpha=0.5,label="versicolor")
plt.hist(data=sub_data["virginica"],x="petal_length",bins=bins,alpha=0.5,label="virginica")

plt.legend()

plt.show()
```

```python

names = ["setosa","versicolor","virginica"]
bins = [i/3 for i in range(25)] #np.linspace(0,8,25)

col = "petal_length"

pars = {"alpha":0.5,"bins":bins,"x":col}

plt.figure(figsize=(8,4)) ## commencer image avec taille (8,4) pouces

for n in names:
    ### histogram
    df_n = sub_data[n]
    
    plt.hist(data=df_n, **pars, label=n )

    ### vlines
    for i,spe in enumerate(GB1["species"]):
        if spe==n:
            mean_n = GB1[col][i]
    
    plt.vlines(mean_n, # x-pos
               ymin=0, 
               ymax=40,
               color='black', 
               label = f'µ {n}: {mean_n}',
               alpha = 0.3
              )
    
    #plt.show()   

plt.legend(ncol=3, loc="upper center")
plt.show()
```

```python
plt.title('mon titre') ## titre graphique
plt.plot(x, y2
         ,linestyle="--" ## style de ligne: tiret
         ,c="red"  ## c=couleur
         ,label='y2'
         ,alpha=0.5
        ) 
### afficher legende
plt.legend() ## ce place automatiquement , le plus idéal

### nommer les axes
plt.xlabel('temps')
plt.ylabel('hauteur')

### modif des val min-max des axes
plt.axis([-3,10,-1,1]) ## [xmin, xmax,ymin, ymax]

plt.show()
```

```python
dict_color= {"setosa":"steelblue","versicolor":"orange","virginica":"purple"}

iris["color"]= [dict_color[spe] for spe in iris["species"]]


plt.scatter(
    data = iris
    ,x="petal_length"
    ,y="petal_width"
    ,c="color"
    ,alpha= 0.8
)
plt.legend() ## je n'ai qu'une seule "couche"
plt.xlabel("petal_length")
plt.ylabel("petal_width")
plt.show()
```

```python
### alternative à la syntaxe suivante

### taille 
plt.figure(figsize=(15, 5)) # taille en pouces

### super titre
plt.suptitle('GRAND TITRE')

### je construit un sous-graphique, 1 row, 2 cols, 1er graphique
### toutes les modif vont s'appliquer à ce sous-graph
plt.subplot(1,2,1) 

plt.title('local') ## titre sous-graphique
plt.plot(x, y2
         ,linestyle="--" ## style de ligne: tiret
         ,c="red"  ## c=couleur
         ,label='y2'
        ) 
### afficher legende
plt.legend() ## ce place automatiquement , le plus idéal

### nommer les axes
plt.xlabel('temps')
plt.ylabel('hauteur')

### modif des val min-max des axes
plt.axis([-3,10,0,1]) ## [xmin, xmax,ymin, ymax]

### definir le 2ième sous-graphique
plt.subplot(1,2,2)
plt.title('local2')
plt.bar(data=GB1, x="species",height="petal_length")


### utiliser le tight_layout== ajouter padding pour eviter le chevauchement des graphiques
plt.tight_layout() 

### sauvegarder l'image en pdf (ou autre)
plt.savefig("img/demo.pdf") ## png, ...
plt.show() # toujours la dernière étape
```

### Style "Orienté Objet"

```python
## fig= full graph
### ax = indexation sous-graphiques (emplacement)
### nrows = nbr lines de sous graphs
### cols = idem en colonnes

x = poly["x"]
y1 = poly["y1"]
y2 = poly["y2"]

### créer un graphique avec des caracteristiques

fig, ax = plt.subplots(nrows=2,ncols=1,figsize=(8, 8)) ## ax= emplacement de graphique (=grille)

ax[0].plot(x, y1, ":", label='y1')
ax[0].plot(x, y2
         ,linestyle="--" ## style de ligne: tiret
         ,c="red"  ## c=couleur
         ,label='y2'
        )

ax[1].bar(data=GB1, x="species",height="petal_length")

ax[0].grid(which='both', color='grey', linewidth=1, linestyle='-', alpha=0.2)


plt.show()
```

```python
fig,ax = plt.subplots()

plt.bar(data=GB1, x="species",height="petal_length")

for bars in ax.containers:
    ax.bar_label(bars) # afficher la valeur d'une barre

plt.xticks(rotation=45)
plt.show()
```
