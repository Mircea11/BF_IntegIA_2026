Créez une fonction valider_mot_de_passe() qui prend un mot de passe en entrée et vérifie s'il répond à certains critères de complexité (longueur minimale, présence de chiffres, de lettres majuscules et minuscules, de caractères spéciaux, etc.). La fonction devrait renvoyer True si le mot de passe est valide et False sinon

````python
def valider_mot_de_passe(mot):
    #longueur minimale, 
    if len(str(mot)) < 7:
        return  False, "longueur minimale"
        # présence de chiffres, 
    elif not any(c.isdigit() for c in mot):
        return  False, "présence de chiffres"
    # de lettres majuscules et minuscules,
    elif not any(c.isupper() for c in mot):
        return  False, "lettres majuscules"
    elif not any(c.islower() for c in mot):
        return  False, "lettres minuscules"
    # de caractères spéciaux
    elif not any(not c.isalnum() for c in mot):
        return False, "caractères spéciaux"
    # return  False
    else:
        return True

print(valider_mot_de_passe("Bonjour123!"))
print(valider_mot_de_passe("123"))
print(valider_mot_de_passe(123))
print(valider_mot_de_passe("Bonjour123"))
print(valider_mot_de_passe("Bonjour!"))
print(valider_mot_de_passe("onjour123!"))
print(valider_mot_de_passe("B123!"))

    
````
