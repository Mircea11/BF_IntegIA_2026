````python
## 5. Jeu du morpion (Tic-Tac-Toe, OXO)
import random

#generating grid
def create_grid():
    grille = {}
    for i in ['a','b','c']:
        for j in range(1,4):
            grille[(i,j)] = None
    return (grille)
#intro_coordonnes_human
def introduire_coordonnees(grille):
    
    # Demander la ligne
    l = input("Introduisez la coordonnée de ligne A, B ou C : ").strip().lower()

    while l not in ["a", "b", "c"]:
        print("La ligne doit être A, B ou C.")
        l = input("Introduisez la coordonnée de ligne A, B ou C : ").strip().lower()

    # Demander la colonne
    n = int(input("Introduisez la coordonnée de colonne 1, 2 ou 3 : "))

    while n not in range(1, 4):
        print("La colonne doit être entre 1 et 3.")
        n = int(input("Introduisez la coordonnée de colonne 1, 2 ou 3 : "))

    print("Coordonnées choisies :", l, n)

    # Vérifier si la case est vide
    if grille[(l, n)] is None:
        grille[(l, n)] = "x"
        change = True
    else:
        print("Cette case est déjà occupée.")
        change = False
        

    return grille, change

#intro_coordonees_rand

def introduire_coordonnees_rand(grille):    
    # Genere la ligne
    
    l_r = random.choice(["a", "b", "c"])
    n_r = random.choice([1,2,3])
    
    print("Coordonnées choisies :", l_r, n_r)

    # Vérifier si la case est vide
    if grille[(l_r, n_r)] is None:
        grille[(l_r, n_r)] = "o"
        change = True
    else:
        change = False
        
    return grille, change

def check_win(grille):
    combinaisons_gagnantes = [
        [("a", 1), ("a", 2), ("a", 3)],
        [("b", 1), ("b", 2), ("b", 3)],
        [("c", 1), ("c", 2), ("c", 3)],

        [("a", 1), ("b", 1), ("c", 1)],
        [("a", 2), ("b", 2), ("c", 2)],
        [("a", 3), ("b", 3), ("c", 3)],

        [("a", 1), ("b", 2), ("c", 3)],
        [("a", 3), ("b", 2), ("c", 1)]
    ]

    for combinaison in combinaisons_gagnantes:
        case1, case2, case3 = combinaison

        if grille[case1] == grille[case2] == grille[case3] and grille[case1] is not None:
            gagnant = grille[case1]
            return True, gagnant

    return False, None


#Main
grille = create_grid()
win, gagnant = check_win(grille)

while not win:
    # player move
    grille, change = introduire_coordonnees(grille)

    # computer move
    grille, change = introduire_coordonnees_rand(grille)

    # check again
    win, gagnant = check_win(grille)

print(f"Le gagnant est : {gagnant}")
````
