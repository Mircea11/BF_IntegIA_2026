```python
a = [1,2,3]
b = ["a","b","c"]
c = []

def fonction_concat_liste(a,b):
    calc = a+b
    return calc

def procedure_concat_liste(a,b,result):
    result.extend(a+b)
    #return None
```

```python
d = fonction_concat_liste(a,b)
print(d)

print("-"*50)

print(c)
res = procedure_concat_liste(a,b,c)
print(c)
print(res)
```

## bonus pour des immutables
- utiliser le mot clé "global"



```python
x = 1
y = 2
z = 0

def fonction_somme(x,y):
    return x+y

def procedure_somme_z(x,y):
    global z

    z = x+y


w = fonction_somme(x,y)
print(w)
print("-"*50)


print(z)
procedure_somme_z(x,y)
print(z)
```

```python
def ma_fonction_avec_documentation(x:str, y:str="world") -> str:
    """documentation textuelle de la fonction"""
    if type(x)!=str:
        return None
        
    msg:str = x + y
    
    return msg#.capitalize()
```

```python
ma_fonction_avec_documentation(x="Hello")
```

```python
ma_fonction_avec_documentation(x=1,y=2)
```
