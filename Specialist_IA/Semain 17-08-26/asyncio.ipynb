on peut indique a asyncio qu'un ensemble de coroutine doive toute etre terminee avant de continue l'execution de code. ceci est possible grace a **asyncio.gather()**

```python
async def meteo():
    
    await asyncio.sleep(2)
    return "17C"
async def actualites():
    await asyncio.sleep(1)
    return ["Pythin 3.15 est sorti", " Incendie de hautes-fagnes est maitrise"]

async def main_gather():
# attente de la fin de toute coroutines dans le 
    resultats = await asyncio.gather(meteo(), actualites())

    resultat_meteo = resultats[0]
    resultat_actus = resultats[1]
    print(f" Meteo recue: il fait {resultat_meteo}")
    print(f" ACtualite recu: {resultat_actus}")


await main_gather()

```

Dans cerain cas, nos taches sont servir a recupere lamem ressources mais san la certitude d'avoire une resultat et /ou des vitesse differents. 
pour ce faire, nous employeront *asyncio.wait()**.



```python
import asyncio
async def actualites_arte():
    
    await asyncio.sleep(3)
    return ["Pythin 3.15 est sorti", " Incendie de hautes-fagnes est maitrise"]

async def actualites_lesoir():
    
    await asyncio.sleep(5)
    return ["Pythin 3.15 est sorti", " Incendie de hautes-fagnes est maitrise"]

async def actualites_eX_twitter():
    
    await asyncio.sleep(10)
    return ["Pythin 3.15 est sorti", " Incendie de hautes-fagnes est maitrise"]

async def main_wait():
    taches = [
        asyncio.create_task(actualites_arte()),
        asyncio.create_task(actualites_lesoir()),
        asyncio.create_task(actualites_eX_twitter()),
    ]

    faites, encours = await asyncio.wait(taches, return_when=asyncio.FIRST_COMPLETED)

    tache_complete = faites.pop()
    print(tache_complete.result())

    #for tache in encours:
    #    tache.cancel()

await main_wait()
```

```python
import asyncio

semaphore = asyncio.Semaphore(5)
async def tache(n):
    async with semaphore:
        print(f"Tache {n} commence")
        await asyncio.sleep(2)
        print(f"Tache {n} se termmine")
async def main_semaphore():
    await asyncio.gather(*(tache(i) for i in range(10)))

await main_semaphore()
```
