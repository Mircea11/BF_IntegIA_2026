## 1. demo via "open meteo" 
API call on "open meteo"

```python
### example:
### doc url: https://open-meteo.com/
### here : max 10_000 call per IP address, because it's free: don't be abusive 


### built url from documentation: 
### try in browser -->> return json
url = f"https://api.open-meteo.com/v1/forecast?latitude={52.52}&longitude={13.41}&hourly=temperature_2m"


### the values associated to "latitude" and "longitude" are parameters => changeable with for loop ! 
```

```python
response = req.get(url)
print (response.status_code) ## 200 = OK
```

### pass parameters inside get (instead of directly in url):




```python
url_base= "https://api.open-meteo.com/v1/forecast?"
params = {"latitude":{52.52},"longitude":{13.41},"hourly":"temperature_2m"}

response = req.get(url=url_base,params=params)
response.status_code
```

```python
response.json().keys()
```

## demo: wikipedia

* for each website, you need to check their robot policy (terms of use)
* here, for wikipedia example:
    * https://wikitech.wikimedia.org/wiki/Robot_policy 
    * https://foundation.wikimedia.org/wiki/Policy:Wikimedia_Foundation_User-Agent_Policy

* for any bot you make, it is good (=polite) practice to add a "User-Agent" metadata to your request
    * for wikipedia,this is required. Otherwise, the bot cannot go there. 

```python
url = "https://en.wikipedia.org/wiki/List_of_countries_by_GDP_(PPP)_per_capita"
headers= {"User-Agent":"demo-request-bruxelles-formation"}
response_wiki = req.get(url=url,headers=headers)

response_wiki.status_code
```

```python
response_wiki.headers["content-type"]
```

```python
response_wiki.encoding = "utf-8"
response_wiki.text[:100]
```

```python
with open("save_page.html","w",encoding="utf-8") as f:
    f.write(response_wiki.text)
```

## GET vs POST

* GET: fetch data that anyone can see
* POST: send data, that may change the server behavior (identification, CRUD, ...)

(https://mockerapi.com/)
