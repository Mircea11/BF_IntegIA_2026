Écrivez un programme Python convertissant un nombre de secondes en jours, heures, minutes et seconds 
correspondants. Exemple : 4561 secondes correspondent à 0 jour, 1 heure, 16 minutes et 1 seconde. Réfléchissez à la 
méthode à utiliser, puis réalisez l'algorithme en Python. Testez-le ensuite pour vérifier sa validité.

```python
Nombre_seconds = int(input("Entrez un nombre de secondes : "))
if Nombre_seconds < 0:
    print("Veuillez entrer un nombre de secondes positif.")
else:
    Nombre_jours = Nombre_seconds // 86400
    Nombre_heurs = (Nombre_seconds - (Nombre_jours * 86400)) // 3600
    Nombre_minutes = (Nombre_seconds - (Nombre_heurs*3600)- (Nombre_jours * 86400)) // 60
    Nombre_secondes = Nombre_seconds - Nombre_heurs*3600 - Nombre_minutes*60 - Nombre_jours*86400
print(f"{Nombre_seconds} secondes correspondent à {Nombre_jours} jours, {Nombre_heurs} heures, {Nombre_minutes} minutes et {Nombre_secondes} secondes")
```

 Créez un programme qui gère les commandes de café en fonction des différentes options telles que la taille, le type de 
café, les extras, etc. Utilisez des correspondances pour traiter chaque option et calculer le prix total de la commande

```python
right_choice = False
while right_choice != True:
    type_de_cafe = input("Entrez le type de café que vous souhaitez (expresso/cappuccino/latte/individuel) : ")
    Case = type_de_cafe.lower()
    if Case == "expresso":
        print("Vous avez choisi un expresso.")  
        cost = 2.5
        right_choice = True
    elif Case == "cappuccino":
        print("Vous avez choisi un cappuccino.")
        cost = 3.0
        right_choice = True
    elif Case == "latte":
        print("Vous avez choisi un latte.")
        cost = 3.5
        right_choice = True
    elif Case == "individuel":
        print("Type de café individuel.")
        cost = int(input("Entrez le prix du café que vous souhaitez : "))
        right_choice = True
    else:  
        print("Type de café non reconnu.")
        cost = 0.0
        right_choice = False

    taille_de_cafe = input("Entrez la taille de café que vous souhaitez petit, moyen, grand: ")
    while taille_de_cafe.lower() not in ["petit", "moyen", "grand"]:
        print("Taille de café non reconnue. Veuillez entrer 'petit', 'moyen' ou 'grand'.")
        taille_de_cafe = input("Entrez la taille de café que vous souhaitez : ")
    if taille_de_cafe.lower() == "petit":
        cost += 0.0
    elif taille_de_cafe.lower() == "moyen":
        cost += 0.5
    elif taille_de_cafe.lower() == "grand":
        cost += 1.0

    extra = input("Souhaitez-vous ajouter un extra (oui/non) ? : ")
    while extra.lower() not in ["oui", "non"]:
        print("Réponse non reconnue. Veuillez entrer 'oui' ou 'non'.")
        extra = input("Souhaitez-vous ajouter un extra (oui/non) ? : ") 
        if extra.lower() == "oui":
            cost += 0.5
        elif extra.lower() == "non":
            cost += 0.0 
print(f"Le coût total de votre café est de {cost} euros.")
```

 Créez un programme qui génère un nombre aléatoire (import random) et permet à l'utilisateur de deviner ce nombre. 
Utilisez des correspondances pour comparer la devinette de l'utilisateur avec le nombre généré et fournir des indices

```python
import random
secret_number = int(random.randint(1, 100))
check = False
attempt = 0
while check == False:
    guess = int(input("Devinez le nombre secret entre 1 et 100 : "))
    
    # print(type(guess))
    # print(type(secret_number))
    # print(secret_number)
    if guess < 1 or guess > 100:
        print("Veuillez entrer un nombre entre 1 et 100.")
        attempt += 1
    elif guess < secret_number:
        print("Le nombre que vous avez entré est trop petit.")
        attempt += 1
    elif guess > secret_number:
        print("Le nombre que vous avez entré est trop grand.")
        attempt += 1
    elif guess == secret_number:
        print(f"Félicitations ! Vous avez deviné le nombre secret en {attempt} tentatives.")
        check =True
```
