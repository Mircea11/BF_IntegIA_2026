```python
# Températures à tester
temperatures = (0.1, 0.5, 1.5)

# Liste pour conserver les résultats de performance
timings = []

# Tester chaque température
for t in temperatures:

    r = client.chat.completions.create(
        model="local",
        messages=[
            {
                "role": "user",
                "content": f"Résume-moi ce texte :\n<document>\n{big_text}\n</document>"
            }
        ],
        temperature=t,
        max_tokens=2000
    )

    # Récupérer les mesures de performance
    metric = r.model_extra["timings"].copy()

    # Ajouter la température utilisée
    metric["temperature"] = t

    # Enregistrer le résultat
    timings.append(metric)

    # Afficher le résultat pour cette température
    print(f"\n===== TEMPÉRATURE : {t} =====")

    print("\nTexte généré :")
    print(r.choices[0].message.content)

    print("\nPerformances :")
    print(metric)
```
