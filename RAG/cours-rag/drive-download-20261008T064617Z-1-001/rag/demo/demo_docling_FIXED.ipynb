```python
files = sorted(p for p in CORPUS.rglob("*") if p.is_file())
export_folder = PROJECT_ROOT / "export_folder"
export_folder.mkdir(parents=True, exist_ok=True)

doc_extraction = {}
for f in files:
    doc = converter.convert(f)
    key = f.relative_to(CORPUS).as_posix()
    doc_extraction[key] = {
        'status': doc.status.value,
        'document': doc.document
    }

    safe_name = key.replace("/", "__")
    doc.document.save_as_markdown(export_folder / f'{safe_name}.md')
    doc.document.save_as_json(export_folder / f'{safe_name}.json')

doc_extraction
```

```python
import subprocess
import sys

from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import PdfPipelineOptions, AcceleratorOptions
from docling.document_converter import DocumentConverter, PdfFormatOption

# The Python environment, model repository and source corpus intentionally live in different locations.
# If the standard Docling model bundle is incomplete, download it once into MODEL_ROOT.
layout_model = MODEL_ROOT / "docling-project--docling-layout-heron"

if not layout_model.is_dir():
    docling_tools = Path(sys.executable).parent / "docling-tools"
    if not docling_tools.is_file():
        raise FileNotFoundError(f"docling-tools introuvable dans le venv actif: {docling_tools}")

    print("Modèles Docling incomplets. Téléchargement dans:", MODEL_ROOT)
    subprocess.run(
        [
            str(docling_tools),
            "models", "download",
            "--rapidocr-backend-lang", "torch:ch",
            "-o", str(MODEL_ROOT),
        ],
        check=True,
    )

options = PdfPipelineOptions(
    artifacts_path=MODEL_ROOT,
    accelerator_options=AcceleratorOptions(device='cpu', num_threads=6),
    generate_page_images=True,
)

converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=options)
    }
)

pdf_files = sorted(CORPUS.glob('*.pdf'))
if not pdf_files:
    raise FileNotFoundError(f"Aucun PDF trouvé dans {CORPUS}")

r = converter.convert(pdf_files[0])
print("Converted:", pdf_files[0].name, "->", r.status.value)
```
