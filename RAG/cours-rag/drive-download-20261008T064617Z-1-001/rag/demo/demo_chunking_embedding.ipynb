```python
import numpy as np
import scipy

import pathlib

from docling_core.transforms.chunker.line_chunker import LineBasedTokenChunker
from docling_core.transforms.chunker.hierarchical_chunker import HierarchicalChunker
from docling_core.transforms.chunker.hybrid_chunker import HybridChunker

from transformers import AutoTokenizer
from docling_core.transforms.chunker.tokenizer.huggingface import HuggingFaceTokenizer
from docling_core.types.doc import DoclingDocument
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    "/home/bm/models/rag/embedder/bge-m3",
    local_files_only=True
)

tokenizer
```

```python
#doc = DoclingDocument.load_from_json('export/intrantet_actualities.html.json')

doc = DoclingDocument.load_from_json(
    "export_folder/intranet_actualites.html.json"
)

doc
docling_tokenizer = HuggingFaceTokenizer(tokenizer=tokenizer, max_tokens=32)



line_chunker = LineBasedTokenChunker(tokenizer=docling_tokenizer)

chunks = line_chunker.chunk(doc)

for c in chunks:
    print (c.text)
    print ('####################################################3')
```
