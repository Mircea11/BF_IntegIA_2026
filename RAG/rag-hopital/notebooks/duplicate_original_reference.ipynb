```python
import re
from unidecode import unidecode
```

```python
text1 = "Harrier est une modele d'embedding performant. "
text2 = "Harrier est une modele d'embedding performant. "
text3 = "Harrier est une modele performant d'embedding . "
```

```python
def normalize(text):
    return ' '.join(re.findall(r'\w+', unidecode(text.lower())))

normalized_text1 = normalize(text1)
normalized_text2 = normalize(text2)

print(normalized_text1)
print(normalized_text2)
```

```python
from datasketch import MinHash, MinHashLSH

text1 = "Harrier est une modele d'embedding performant. "
text2 = "Harrier est une modele d'embedding performant. "
text3 = "Harrier est une modele performant d'embedding . "

hashs = MinHash.bulk(
    [

    [s.encode('utf-8') for s in shingle(text1)],
    [s.encode('utf-8') for s in shingle(text2)],
    [s.encode('utf-8') for s in shingle(text3)]

    ]
    , num_perm=128
)

hashs[0].hashvalues

lsh = MinHashLSH(threshold=0.5, num_perm=128)   

for i, h in enumerate(hashs):
    lsh.insert(f"text{i}", h)

lsh.query(hashs[2])
```
