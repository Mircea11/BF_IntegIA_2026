#### 18 Afficher l'évolution(en valeur absolue et en %), d'année en année, du nombre d'offenses faites aux femmes depuis 2010

<br>

```python
# This block displays the distinct values of the victim sex column.
valeur_sex = data.select(
    pl.col("vic_sex").unique().sort()
).collect()

valeur_sex
```

```python

# This block counts offenses against female victims by year since 2010.
offenses_femmes_par_annee = (
    data
    .filter(
        (pl.col("vic_sex") == "F")
        & (pl.col("cmplnt_fr_dt").dt.year() >= 2010)
    )
    .with_columns(
        pl.col("cmplnt_fr_dt").dt.year().alias("annee")
    )
    .group_by("annee")
    .agg(
        pl.len().cast(pl.Int64).alias("nombre_offenses")
    )
    .sort("annee")
)

offenses_femmes_par_annee.collect()
```

```python


# This block calculates the absolute year-on-year change.
offenses_femmes_evolution = (
    offenses_femmes_par_annee
    .with_columns(
        (
            pl.col("nombre_offenses")
            - pl.col("nombre_offenses").shift(1)
        ).alias("evolution_absolue")
    )
)

offenses_femmes_evolution.collect()
```

```python
# This block calculates the year-on-year percentage change.
offenses_femmes_evolution = (
    offenses_femmes_evolution
    .with_columns(
        (
            pl.col("evolution_absolue")
            / pl.col("nombre_offenses").shift(1)
            * 100
        )
        .round(2)
        .alias("evolution_pourcentage")
    )
)

offenses_femmes_evolution.collect()
```

```python
pl.col("cmplnt_fr_dt").dt.hour()
# This block adds a day-period column based on the complaint start hour.
data = data.with_columns(
    pl.when(
        (pl.col("cmplnt_fr_dt").dt.hour() >= 22)
        | (pl.col("cmplnt_fr_dt").dt.hour() < 5)
    )
    .then(pl.lit("NIGHT"))

    .when(
        (pl.col("cmplnt_fr_dt").dt.hour() >= 5)
        & (pl.col("cmplnt_fr_dt").dt.hour() < 19)
    )
    .then(pl.lit("DAY"))

    .otherwise(pl.lit("EVENING"))

    .alias("periode_journee")
)
```

```python
data.select(
    "cmplnt_fr_dt",
    "periode_journee"
).head(10).collect()
```

```python
# This block counts records by day period and offense category.
infractions_par_periode = (
    data
    .group_by(
        "periode_journee",
        "law_cat_cd"
    )
    .agg(
        pl.len().alias("nombre_enregistrements")
    )
    .sort(
        ["periode_journee", "nombre_enregistrements"],
        descending=[False, True]
    )
    .collect()
)

infractions_par_periode
```

```python
# This block creates the day-period column.
valeur_day_time = (
    data
    .with_columns(
        pl.when(
            (pl.col("cmplnt_fr_dt").dt.hour() >= 22)
            | (pl.col("cmplnt_fr_dt").dt.hour() < 5)
        )
        .then(pl.lit("NIGHT"))

        .when(
            (pl.col("cmplnt_fr_dt").dt.hour() >= 5)
            & (pl.col("cmplnt_fr_dt").dt.hour() < 19)
        )
        .then(pl.lit("DAY"))

        .otherwise(pl.lit("EVENING"))
        .alias("periode_journee")
    )
    .select("periode_journee")
    .unique()
    .collect()
)

valeur_day_time
```

#### 23 Proposez un crosstab avec le nombre d'enregistrements par type d'infractions et moment de la journée

Plusieurs remarques:
- nous voudrions également avoir non pas les valeurs absolues mais les ratios. Ces ratios peuvent être calculer de trois façons:
    - en divisant chaque cellule par la somme de sa colonne
    - en divisant chaque cellule par la somme de sa ligne
    - en divisant chaque cellule par le total de la matrice
- l'exercice est plus facile à réaliser si vous passez par Pandas qui propose une solution avec crosstab. Nous voudrions que vous proposiez également une solution sans passer par Pandas.

<br>

```python
# This block collects only the columns needed for the crosstab and converts them to pandas.
df_cross = (
    data
    .select(
        "law_cat_cd",
        "periode_journee"
    )
    .collect()
    .to_pandas()
)
```

```python
import pandas as pd
# This block creates the absolute-count crosstab.
crosstab_count = pd.crosstab(
    df_cross["law_cat_cd"],
    df_cross["periode_journee"]
)

crosstab_count
```

```python
# This block calculates row-normalized ratios.
crosstab_ligne = pd.crosstab(
    df_cross["law_cat_cd"],
    df_cross["periode_journee"],
    normalize="index"
)

crosstab_ligne
```

```python
# This block calculates column-normalized ratios.
crosstab_colonne = pd.crosstab(
    df_cross["law_cat_cd"],
    df_cross["periode_journee"],
    normalize="columns"
)

crosstab_colonne
```

```python
# This block calculates ratios relative to the whole table.
crosstab_total = pd.crosstab(
    df_cross["law_cat_cd"],
    df_cross["periode_journee"],
    normalize="all"
)

crosstab_total
```

```python
# This block counts records for each offense category and day period.
cross_long = (
    data
    .group_by(
        "law_cat_cd",
        "periode_journee"
    )
    .agg(
        pl.len().alias("nombre_enregistrements")
    )
)
```

```python
# This block pivots the long-format counts into a crosstab.
cross_polars = (
    cross_long
    .collect()
    .pivot(
        values="nombre_enregistrements",
        index="law_cat_cd",
        on="periode_journee"
    )
)

cross_polars
```

```python
import pandas as pd
# This block creates the absolute-count crosstab.
crosstab_count = pd.crosstab(
    df_cross["law_cat_cd"],
    df_cross["periode_journee"]
)

crosstab_count
```

```python
# This block calculates ratios across each row.
cross_ratio_ligne = (
    cross_polars
    .with_columns(
        (
            pl.col("DAY")
            / pl.sum_horizontal(
                pl.col("DAY"),
                pl.col("EVENING"),
                pl.col("NIGHT")
            )
        ).alias("DAY_ratio"),

        (
            pl.col("EVENING")
            / pl.sum_horizontal(
                pl.col("DAY"),
                pl.col("EVENING"),
                pl.col("NIGHT")
            )
        ).alias("EVENING_ratio"),

        (
            pl.col("NIGHT")
            / pl.sum_horizontal(
                pl.col("DAY"),
                pl.col("EVENING"),
                pl.col("NIGHT")
            )
        ).alias("NIGHT_ratio")
    )
)

cross_ratio_ligne
```

```python
# This block calculates ratios relative to each column total.
cross_ratio_colonne = (
    cross_polars
    .with_columns(
        (pl.col("DAY") / pl.col("DAY").sum()).alias("DAY_ratio"),
        (pl.col("EVENING") / pl.col("EVENING").sum()).alias("EVENING_ratio"),
        (pl.col("NIGHT") / pl.col("NIGHT").sum()).alias("NIGHT_ratio")
    )
)

cross_ratio_colonne
```

```python
# This block calculates the grand total of all crosstab cells.
total_general = (
    cross_polars
    .select(
        pl.sum_horizontal(
            pl.col("DAY"),
            pl.col("EVENING"),
            pl.col("NIGHT")
        ).sum()
    )
    .item()
)
```

```python
# This block calculates each cell's share of the whole matrix.
cross_ratio_total = (
    cross_polars
    .with_columns(
        (pl.col("DAY") / total_general).alias("DAY_ratio"),
        (pl.col("EVENING") / total_general).alias("EVENING_ratio"),
        (pl.col("NIGHT") / total_general).alias("NIGHT_ratio")
    )
)

cross_ratio_total
```
