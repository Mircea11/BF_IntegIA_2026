```python
# This block combines complaint start date + time, complaint end date + time,
# and converts the report date into a real Date column.
data_new_time = df.with_columns(
    pl.concat_str(
        [pl.col("cmplnt_fr_dt"), pl.col("cmplnt_fr_tm")],
        separator=" ",
        ignore_nulls=False
    )
    .str.strptime(pl.Datetime, format=r"%m/%d/%Y %H:%M:%S", strict=False)
    .alias("cmplnt_fr_datetime"),

    pl.concat_str(
        [pl.col("cmplnt_to_dt"), pl.col("cmplnt_to_tm")],
        separator=" ",
        ignore_nulls=False
    )
    .str.strptime(pl.Datetime, format=r"%m/%d/%Y %H:%M:%S", strict=False)
    .alias("cmplnt_to_datetime"),

    pl.col("rpt_dt")
    .str.strptime(pl.Date, format=r"%m/%d/%Y", strict=False)
    .alias("rpt_date")
)

# This block verifies the new datetime/date columns.
print(data_new_time.select(
    "cmplnt_fr_dt",
    "cmplnt_fr_tm",
    "cmplnt_fr_datetime",
    "cmplnt_to_dt",
    "cmplnt_to_tm",
    "cmplnt_to_datetime",
    "rpt_dt",
    "rpt_date"
).head(2))


print(data_new_time.head(5))


# This block removes the two time columns because they were already combined into datetime columns.
data_new_time_less = data_new_time.drop(["cmplnt_fr_tm", "cmplnt_to_tm", "cmplnt_fr_dt", "cmplnt_to_dt"])

print(data_new_time_less.head(3))
```

#### 5 Vérifiez si les colonnes en Int64 peuvent être converties en Int32 afin de gagner en espace mémoire

<br>

```python
# This block defines the Int32 limits and reads the current LazyFrame schema.
INT32_MIN = -2_147_483_648
INT32_MAX = 2_147_483_647

schema = df.collect_schema()

int64_cols = [
    column_name
    for column_name, dtype in schema.items()
    if dtype == pl.Int64
]

print("Colonnes Int64 trouvées :", int64_cols)


# This block checks whether Int64 columns exist.
# If yes, it verifies whether they can safely be converted to Int32.
# If no, it skips the conversion without crashing.
if len(int64_cols) == 0:
    print("Aucune colonne Int64 trouvée. Rien à convertir.")

    int64_check = pl.DataFrame({
        "column": pl.Series([], dtype=pl.String),
        "min_value": pl.Series([], dtype=pl.Int64),
        "max_value": pl.Series([], dtype=pl.Int64),
        "can_cast_to_int32": pl.Series([], dtype=pl.Boolean),
    })

    cols_to_cast = []

else:
    stats_wide = df.select(
        *[pl.col(col).min().alias(f"{col}__min") for col in int64_cols],
        *[pl.col(col).max().alias(f"{col}__max") for col in int64_cols],
    )
    stats_row = stats_wide.row(0, named=True)

    int64_check = pl.DataFrame({
        "column": int64_cols,
        "min_value": [stats_row[f"{col}__min"] for col in int64_cols],
        "max_value": [stats_row[f"{col}__max"] for col in int64_cols],
    }).with_columns(
        (
            (pl.col("min_value") >= INT32_MIN)
            & (pl.col("max_value") <= INT32_MAX)
        ).alias("can_cast_to_int32")
    )

    cols_to_cast = (
        int64_check
        .filter(pl.col("can_cast_to_int32"))
        .get_column("column")
        .to_list()
    )

print(int64_check)
print("Colonnes à convertir :", cols_to_cast)

# This block converts only the compatible Int64 columns into Int32.
# If there are no columns to convert, it safely skips the operation.
if len(cols_to_cast) > 0:
    df = df.with_columns(
        pl.col(cols_to_cast).cast(pl.Int32)
    )
    print("Conversion effectuée :", cols_to_cast)
else:
    print("Aucune conversion effectuée.")

    # This block verifies the schema after the conversion check.
print(df.collect_schema())
print(df.shape)
```
