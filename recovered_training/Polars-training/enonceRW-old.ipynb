#### 12  En moyenne, quel est le temps écoulé entre le début d'un rapport et de sa fin ?

<br>

```python
# This block calculates one scalar value: the average elapsed time in days.
temps_moyen_jours = (
    data
    .filter(
        pl.col("cmplnt_fr_dt").is_not_null()
        & pl.col("cmplnt_to_dt").is_not_null()
    )
    .with_columns(
        (pl.col("cmplnt_to_dt") - pl.col("cmplnt_fr_dt")).alias("temps_ecoule")
    )
    .filter(
        pl.col("temps_ecoule").dt.total_seconds() >= 0
    )
    .select(
        (pl.col("temps_ecoule").dt.total_seconds().mean() / 86400)
        .round(2)
        .alias("temps_moyen_jours")
    )
    .collect()
    .item()
)

temps_moyen_jours
```

#### 13 Est-ce que ce temps moyen est similaire entre les différents quartiers ? Triez par ordre décroissant et afficher également le nombre d'offenses recensées.

<br>

```python
# This block calculates the average elapsed time by borough
# and counts the number of offenses in each borough.
temps_moyen_par_quartier = (
    data
    .filter(
        pl.col("boro_nm").is_not_null()
        & pl.col("cmplnt_fr_dt").is_not_null()
        & pl.col("cmplnt_to_dt").is_not_null()
    )
    .with_columns(
        (pl.col("cmplnt_to_dt") - pl.col("cmplnt_fr_dt")).alias("temps_ecoule")
    )
    .filter(
        pl.col("temps_ecoule").dt.total_seconds() >= 0
    )
    .group_by("boro_nm")
    .agg(
        pl.len().alias("nombre_offenses"),
        (
            pl.col("temps_ecoule").dt.total_seconds().mean() / 86400
        )
        .round(2)
        .alias("temps_moyen_jours")
    )
    .sort("temps_moyen_jours", descending=True)
    .collect()
)

temps_moyen_par_quartier
```

#### 14 Affichez par année et ce depuis 2010 la somme cumulée du nombre d'offenses de catégorie *Felony*

**BONUS**: Pouvez-vous constater une accélération ?

<br>

```python
felony_cumulative_by_year = (
    data
    .filter(
         pl.col("cmplnt_fr_dt").is_not_null()
        & pl.col("law_cat_cd").is_not_null()

    )
    .with_columns(
        pl.col("cmplnt_fr_dt").dt.year().alias("annee")
        )
        .filter(
        (pl.col("annee") >= 2010)
        & (pl.col("law_cat_cd") == "FELONY")
        ).group_by("annee")
        .agg(
            pl.len().alias("nombre_offenses")
        )
        .sort("annee")
        .with_columns(
            pl.col("nombre_offenses").cum_sum().alias("somme_cumulee_offenses")
        ).collect()

    )
```

```python

felony_cumulative_by_year
```

```python
data.with_columns(
    pct_change_over_year = pl.nombre_offenses
)
```

```python
# This block adds yearly variation and percentage variation
# to check whether Felony offenses are accelerating or slowing down.
felony_acceleration = (
    felony_cumulative_by_year
    .with_columns(
        pl.col("nombre_offenses").cast(pl.Int64)
         )
         .with_columns(
             (
                 pl.col("nombre_offenses") - pl.col("nombre_offenses").shift(1)
         ).alias("variation_annuelle")
         )
         .with_columns(
             (    
                pl.col("variation_annuelle")
                 / pl.col("nombre_offenses").shift(1)
                * 100
        )
        .round(2)
        .alias("variation_pourcentage")
    ).with_columns(
        pl.col("variation_pourcentage")
        .fill_null(0)
        .cum_sum()
        .round(2)
        .alias("somme_variation_pourcentage")
    )
)

felony_acceleration
```

```python
# This block calculates acceleration as the change in annual variation.
felony_acceleration = (
    felony_acceleration
    .with_columns(
        (
            pl.col("variation_annuelle") - pl.col("variation_annuelle").shift(1)
        ).alias("acceleration_annuelle")
    )
)

felony_acceleration
```
