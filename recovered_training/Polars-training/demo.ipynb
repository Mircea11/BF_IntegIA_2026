```python
data.select(

#pl.col('VendorID','tpep_pickup_datetime')#.show_graph() -ne march pas
pl.all()
, (pl.col('trip_distance')/pl.col('passenger_count')).alias('dustance_per_passenger') 
)
```

```python
data.select(

pl.exclude('trip_distance','passenger_count') 
, (pl.col('trip_distance')/pl.col('passenger_count')).alias('dustance_per_passenger') 
)
```

```python
data.select(
    pl.col(pl.Int32)
    , pl.col('tpep_pickup_datetime')
)

data.select(
    cs.contains('datetime')
    , cs.starts_with
    , cs.ends_with
    ).collect()
```

```python
data.select(
    pl.exclude('tpep_dropoff_datetime')
).filter(
    (pl.col('trip_distance') >= 50)
    & (pl.col('trip_distance') < 80)
)#.show_graph()
```

```python
data.select(
    pl.exclude('tpep_dropoff_datetime')
).filter(
    (pl.col('trip_distance') >= 50)
    & (pl.col('trip_distance') < 80)
    | pl.col('VendorID').is_in((2,7))
)#.show_graph()
```

```python
data.select(
    pl.exclude('tpep_dropoff_datetime')
).filter(
    (pl.col('trip_distance') >= 50)
    & (pl.col('trip_distance') < 80)
    | ~(pl.col('VendorID').is_in((2,7)))
)#.show_graph()
```

```python
data.select(
    pl.exclude('tpep_dropoff_datetime')
).filter(
   pl.col('trip_distance').is_between(50,80)
   , pl.col('trip_distance').eq(50) 
   , pl.col('store_and_fwd_flag').str.contains('N')
)#.show_graph()
```

```python
data.select(
    pl.exclude('tpep_dropoff_datetime')
).filter(
   pl.col('trip_distance').is_between(50,80)
   
).collect() 
```

```python
data.select(
    pl.exclude('tpep_dropoff_datetime')
).filter(
   pl.col('trip_distance').is_between(50,80)
   
).cast(
    {
        'trip_distance': pl.Float32
        ,'total_amount': pl.Decimal(18,2)
        ,'passenger_count': pl.UInt8
        }
).head(5).collect() 
```

```python
(
data.with_columns(
    trip_category = pl.when(pl.col('trip_distance') < 0).then(pl.lit('anomalies'))
    .when(pl.col('trip_distance').is_between(0,5)).then(pl.lit('small trip'))
    .otherwise(pl.lit('big trip')))
.select(pl.col('trip_category'))
)
```

```python
(
data
    .select(pl.col('trip_distance'))\
    .with_columns(
    trip_category = pl.when(pl.col('trip_distance') < 0).then(pl.lit('anomalies'))
    .when(pl.col('trip_distance').is_between(0,5)).then(pl.lit('small trip'))
    .otherwise(pl.lit('big trip')))
.select(pl.col('trip_category'))
).collect()
```

```python
(
data
    
    .with_columns(
        trip_category = pl.when(pl.col('trip_distance') < 0).then(pl.lit('anomalies'))
        .when(pl.col('trip_distance').is_between(0,5)).then(pl.lit('small trip'))
        .otherwise(pl.lit('big trip')))
    .group_by('trip_category')
    .agg(
        pl.col('passenger_count').mean()
        , pl.col('passenger_count').mean().alias('mean_passaenger_count')
        , pl.col('passenger_count').median().alias('median_passaenger_count')
        ,pl.col('total_amount').sum()
    )
).collect()
```
